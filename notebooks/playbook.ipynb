```
using System.IO;
using System.Linq;  
using Microsoft.Data.Analysis;

var dataPath = Path.GetFullPath(@"housing-prices.csv"); 
```

```
var df = DataFrame.LoadCsv(dataPath);

df.Display()
```

```
df["ComputedPrice"] = df["HistoricalPrice"].Multiply(2);

df
```

```
df.Filter(df["CurrentPrice"].ElementwiseGreaterThan(200000))
```

```
using Microsoft.ML;

var result = from data in df.Rows 
    select new {Id = data["Id"] as float? ?? default, 
                ComputedPrice = data["ComputedPrice"] as double? ?? default};

result.Display()
```

```
using Microsoft.ML.Data;

var ctx = new MLContext();

var ndf = ctx.Data.LoadFromEnumerable(result);

ndf.ToDataFrame()
```
